# 01 — MDP Formulation
**Week 3 | RL Fundamentals**

A **Markov Decision Process** (MDP) is defined by the 5-tuple:

$$M = (\mathcal{S},\; \mathcal{A},\; \mathcal{R},\; \mathcal{T},\; \gamma)$$

| Symbol | Meaning |
|---|---|
| S | State space |
| A | Action space |
| R(s,a,s') | Reward function |
| T(s,a,s') | Transition function P(s'|s,a) |
| γ ∈ [0,1) | Discount factor |

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
np.random.seed(0)
```

## 1. Constructing a Tiny MDP by Hand
A recycling robot (from Sutton & Barto, Example 3.3):  
- **States**: {high battery, low battery}  
- **Actions**: {search, wait, recharge}  
- **Goal**: maximise cans collected

```python
# States
S = ['high', 'low']
s_idx = {s: i for i, s in enumerate(S)}

# Actions (not all available in every state but we'll handle that via masks)
A = ['search', 'wait', 'recharge']
a_idx = {a: i for i, a in enumerate(A)}

# Transition tensor: T[s, a, s'] = P(s'|s,a)
# and Reward matrix:  R[s, a] = expected reward
alpha = 0.9   # P(stay high | high, search)
beta  = 0.4   # P(stay low  | low,  search)
r_search_high = 4.0
r_search_low  = 2.0
r_wait        = 1.0

T = np.zeros((2, 3, 2))   # (states, actions, next_states)
R = np.full((2, 3), -np.inf)  # -inf for illegal actions

# From HIGH
T[0, 0, 0] = alpha;    T[0, 0, 1] = 1 - alpha;  R[0, 0] = r_search_high  # search
T[0, 1, 0] = 1.0;                                R[0, 1] = r_wait         # wait
# recharge not available from high

# From LOW
T[1, 0, 0] = 1-beta;   T[1, 0, 1] = beta;        R[1, 0] = r_search_low   # search
T[1, 1, 1] = 1.0;                                 R[1, 1] = r_wait         # wait
T[1, 2, 0] = 1.0;                                 R[1, 2] = 0.0            # recharge

print("Transition tensor shape:", T.shape, "  (states × actions × next_states)")
print("\nReward matrix R[s,a]:")
for i, s in enumerate(S):
    for j, a in enumerate(A):
        r = R[i, j]
        print(f"  R[{s},{a}] = {r if r != -np.inf else 'N/A'}")
```

## 2. Simulate One Episode Under a Random Policy

```python
def valid_actions(s):
    if s == 0:  return [0, 1]        # high: search, wait
    else:       return [0, 1, 2]     # low:  search, wait, recharge

def step(s, a):
    next_s = np.random.choice(2, p=T[s, a])
    r = R[s, a]
    return next_s, r

def simulate_episode(start=0, n_steps=20, gamma=0.95):
    s = start
    trajectory = []
    for t in range(n_steps):
        a = np.random.choice(valid_actions(s))
        next_s, r = step(s, a)
        trajectory.append({'t': t, 'state': S[s], 'action': A[a], 'reward': r, 'next': S[next_s]})
        s = next_s
    return trajectory

traj = simulate_episode()
print(f"{'t':>3}  {'State':<8} {'Action':<10} {'Reward':>7}  {'Next':>8}")
print("-" * 45)
for step_ in traj[:10]:
    print(f"{step_['t']:>3}  {step_['state']:<8} {step_['action']:<10} {step_['reward']:>7.1f}  {step_['next']:>8}")
```

## Q1: Frozen Lake 4×4 as an MDP
- States: 16 — one per grid cell (4×4 grid), indexed 0–15.
- Actions: 4 — Left (0), Down (1), Right (2), Up (3).
- Terminal states: 5 — the Goal cell (G at position 15) and the 4 Hole cells (H at positions 5, 7, 11, 12 in the standard layout).
- Reward: +1 on reaching the Goal, 0 everywhere else.
- Transition: Stochastic — due to the slippery surface, the intended action is taken with probability 1/3, and each perpendicular direction is taken --- with probability 1/3 each.

 ## Q2: Noughts and Crosses (Tic-Tac-Toe)
| **Component** | **Definition** |
|--------------|----------------|
| **S** | All valid board configurations — each cell can be **X**, **O**, or empty. At most \(3^9 = 19,683\) configurations, ~5,478 valid reachable states. |
| **A** | Place a mark on any empty cell — up to 9 actions at the start, decreasing as the board fills. |
| **R** | +1 for a win, −1 for a loss, 0 for a draw or any non-terminal step. |
| **T** | Deterministic — placing a mark on cell *i* transitions to exactly one new board state. \(T(s,a,s') = 1\) for the resulting configuration, 0 for all others. |
Terminal states: Any board where a player has three in a row, or all 9 cells are filled (draw).

## Q3 1000 Episodes — Always Search Policy

```python
def simulate_episode_always_search(start=0, n_steps=20):
    s = start
    total_reward = 0
    for _ in range(n_steps):
        a = 0  # always search
        next_s, r = step(s, a)
        total_reward += r  # undiscounted
        s = next_s
    return total_reward

episode_rewards = [simulate_episode_always_search() for _ in range(1000)]

plt.figure(figsize=(8, 3.5))
plt.hist(episode_rewards, bins=30, color='steelblue', edgecolor='white')
plt.xlabel('Total undiscounted reward')
plt.ylabel('Count')
plt.title(f'Recycling Robot — Always Search (1000 episodes, mean={np.mean(episode_rewards):.2f})')
plt.tight_layout(); plt.show()

print(f"Mean reward:   {np.mean(episode_rewards):.2f}")
print(f"Std reward:    {np.std(episode_rewards):.2f}")
print(f"Min reward:    {np.min(episode_rewards):.2f}")
print(f"Max reward:    {np.max(episode_rewards):.2f}")
```
